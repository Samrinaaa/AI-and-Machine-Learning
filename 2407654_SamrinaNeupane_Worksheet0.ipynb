## Data types in Python
### Mutable datatypes
these are the data types that can be changed/modified after they have been created. for eg: list. after a list is created we can add to it remove from it and we can also replace elements using append, remove/pop and insert respectively
### Immutable datatypes
these are the datatypes that cannot be changed/modified after they have been created. any operation that seems to modify said datatype will actually just create a new object instead of changing the original object. for eg: tuple

```python
my_list = [1,2,3,4] #lists are mutable which means we can add and remove from the list. we can also have duplicate values here.
print("original list: ",my_list)
my_list.append(5) #append adds the element to the end of the list
print("New list: ",my_list)
my_list.remove(2) #remove removes the first occurence of the said element from the list. if more than one of the same element only the first one is removed
print("New list: ",my_list)
my_list.insert(1,10) #insert adds the element (10) at index (1) which can be specified
print("New list: ",my_list)
my_list.pop() #this removes the last element of the list.
print("New list: ",my_list)
my_list.pop(2) #we can also specify the index of the element we want to remove using pop. i wrote 2 which will remove the element at index 2 which is 3
print("New list: ",my_list)
```

```python
my_tuple = (1,2,3,4)
print("original tuple: ",my_tuple)
# my_tuple.append(5) trying to add to this will raise an error because tuple are immutable
my_tuple = my_tuple + (5,) #we can add to a tuple by concatinating with another tuple. if number is single we need to add comma after it to make it a tuple else it is just a number
print(my_tuple) #if there are multiple numbers we can write in the parenthesis without comma at the end. eg: (5,6,7) but adding the comma will also work so (5,6,7,) will also work
```

## Common   data types in python
### numeric:
1. int: this represents integer values or simply numbers. so 1, 2, 10, 10001, -1000 all are integer values.
2. float: this represents floating type values or decimal numbers. so 1.2, -1.5, 10000.23 all are float values.
3. complex: represents complex values. so 3+4j
### sequence:
these are the data types that are ordered collection of items. here we can access item/element by index
1. string: this is a collection of characters enclosed within double quotes or single quotes. eg: "Hello world". these are immutable.
2. list: this represents a list. so a collection of items in a list which can be of mixed datatypes. these are mutable datatype.
3. tuple: this represents a tuple which are ordered and immutable collection of items. they can also have mixed datatype but are immutable.
### mapping:
this includes data type that store value in key value pair allowing retreival of data based on keys.
1. dictionary: represents dict(dictionary) which can store multi datatype as values with unique keys. these are mutable.
### set:
sets are unordered collection of unique elements. these are useful for membership testing and eleminating duplicate values. these are mutable. but frozenset are immutable versions of set.
### boolean:
this represents the truth value of an element. this can be true of false.
### special:
1. NoneType: represents abscence of value or null value. eg: None

```python
def length_conversion(length, length_type):
    '''
    Converts units of length from meter to feet or feet to meter and returns the converted value to the user.
    Parameters:
    length (float): A floating point value
    length_type (String): A collection of characters used to get the unit type of length (m or ft)
    '''
    if(length_type=="m"):
        length_in_feet = length*3.28084
        return length_in_feet
    elif(length_type=="ft"):
        length_in_meters = length/3.28084
        return length_in_meters
    else:
        return None

def weight_conversion(weight, weight_type):
    """
    Converts units of weight from kg to lbs or lbs to kg and returns the converted value to the user.
    Parameters:
    weight (float): A floating point value
    weight_type (String): A collection of characters used to get the unit type of weight (kg or lbs)
    """
    if(weight_type=="kg"):
        weight_in_pounds = weight*2.20462
        return weight_in_pounds
    elif(weight_type=="lbs"):
        weight_in_kg = weight/2.20462
        return weight_in_kg
    else:
        return None

def volume_conversion(volume, volume_type):
    '''
    Converts units of volume from liters to gallons or gallons to liters and returns the converted value to the user.
    Parameters:
    volume (float): A floating point value
    volume_type (String): A collection of characters used to get the unit type of volume (l or gal)
    '''
    if(volume_type=="l"):
        volume_in_gallons= volume*0.264172
        return volume_in_gallons
    elif(volume_type=="gal"):
        volume_in_liters = volume/0.264172
        return volume_in_liters
    else:
        return None

get_conversion_type = input("Enter the type of conversion you want: length, weight or volume:") #getting the conversion type from the user
try: #its better if i do the .lower() to the get_conversion_type above while taking input to remove repeating the same thing
    if(get_conversion_type.lower()=="length"): #checking if conversion type is length
        length_type = input("Enter unit of length you want to convert from (m or ft): ").strip().lower() #strip to rid of spaces and whitespaces
        length = float(input(f"Enter length in {length_type}: "))
        converted_length = length_conversion(length, length_type) #getting the converted value by passing arguements to function and getting the returned result
        print(f"Converted length: {converted_length:.2f}")
    elif(get_conversion_type.lower()=="weight"): #checking if conversion type is weight
        weight_type = input("Enter unit of weight you want to convert from (kg or lbs): ").strip().lower() #strip to rid of spaces and whitespaces
        weight = float(input(f"Enter weight in {weight_type}: "))
        converted_weight = weight_conversion(weight, weight_type) #getting the converted value by passing arguements to function and getting the returned result
        print(f"Converted weight: {converted_weight:.2f}")
    elif(get_conversion_type.lower()=="volume"): #checking if conversion type is volume
        volume_type = input("Enter unit of volume you want to convert from (l or gal): ").strip().lower() #strip to rid of spaces and whitespaces
        volume = float(input(f"Enter volume in {volume_type}: "))
        converted_volume = volume_conversion(volume, volume_type) #getting the converted value by passing arguements to function and getting the returned result
        print(f"Converted volume: {converted_volume:.2f}")
    else:
        print("Invalid conversion type")
except Exception as e: #error handling for invalid input, type and so on
    print("Error: ",e)
```

```python
user_choice = input("choose an operator: sum, avg, max or min: ").strip().lower()
user_input = input("Enter the numbers separated by space: ")

user_input_list = [float(x) for x in user_input.split()]

print(f"your input: {user_input_list}")

def sum_func(lst):
    total = 0
    for i in lst:
        total += i
    return total

def avg_func(lst):
    return sum_func(lst) / len(lst)

def max_func(lst):
    maximum = lst[0]
    for i in lst:
        if i > maximum:
            maximum = i
    return maximum

def min_func(lst):
    minimum = lst[0]
    for i in lst:
        if i < minimum:
            minimum = i
    return minimum

try:
    if user_choice == "sum":
        print(f"sum of list is: {sum_func(user_input_list)}")

    elif user_choice == "avg":
        print(f"avg of list is: {avg_func(user_input_list)}")

    elif user_choice == "max":
        print(f"max of list: {max_func(user_input_list)}")

    elif user_choice == "min":
        print(f"min of list: {min_func(user_input_list)}")

    else:
        print("Invalid Operation")

except Exception as e:
    print("Error:", e)
```

```python
def get_sublist(lst,start,end):
    '''
    this function takes user input for list, start slice and end slice to return a new list of sliced list.
    Parameters:
    lst (list): this is a collectin of elements that is input by the user
    start (int): this is a numeric value input by user for start index of slicing
    end (int): this is a numeric value input by the user for the end index of slicing
    '''
    new_lst = lst[start:end+1] #end+1 because slicing end is exclusive so it does not get included so +1 will make it included
    return new_lst

user_input = input("Enter elements of the list separated by commas: ")
user_input_list = [float(x) for x in user_input.split(",")]
get_start_index = int(input("Enter the start index where to start slicing: "))
get_end_index = int(input("Enter the end index where to end slicing: "))
result = get_sublist(user_input_list,get_start_index,get_end_index)
print(user_input_list)
print(f"{result}")
```

```python
def reverse_list(lst):
    '''
    this function returns user input list in a reversed format
    Parameters:
    lst (list): this is a list of items user input
    '''
    return lst[::-1] #so making step -1 makes it go from back to front. so -1 step means move backwards

user_input = input("Enter the list items separated by a space: ")
user_input_list = [float(x) for x in user_input.split()]
result = reverse_list(user_input_list)
print(user_input_list)
print(result)
```

```python
def get_first_n(lst,n):
    '''
    this fucntion returns the first n amount of elements of the list the user specifies
    Parameters:
    lst (list): collection of elements input by the user
    n (int): this is a numeric value input by the user
    '''
    return lst[:n] #here no need for +1 as if n=3 it will print the index 0-1-2 which is three elements but if we wanted to get n=3 meaning till index 3 then i would have to do +1

user_input = input("Enter the list items separated by space: ")
user_input_list = [int(x) for x in user_input.split()]
n_elements = int(input("Enter the number of first n elements you want to see: "))
result = get_first_n(user_input_list,n_elements)
print(user_input_list)
print(result)
```

```python
def get_last_n(lst, n):
    '''
    this function returns the n amount of last elements of the list.
    Parameters:
    lst (list): this is a collection of elements input by the user
    n (int): this is a numeric value input by the user
    '''
    return lst[n+1:] #we want the last n number of elements so n+1 will remove n and start from n+1. so n=2 it will start from index 2 which will give 3 elements but +1 will give the last 2 elements

user_input = input("Enter the list items separated by space: ")
user_input_list = [int(x) for x in user_input.split()]
n_elements = int(input("Enter the n amount from last you want to see: "))
result = get_last_n(user_input_list,n_elements)
print(user_input_list)
print(result)
```

```python
def sum_nested_list(lst):
    total = 0
    for i in lst:
        if isinstance(i, list):   # check if element is a list
            total += sum_nested_list(i)   # recursive call
        else:
            total += i
    return total


nested_list = eval(input("Enter nested list: "))
print(nested_list)

result = sum_nested_list(nested_list)

print("Sum of nested list:", result)
```

```python
def remove_element(lst,elem):
    new_list = []
    for i in lst:
        if isinstance(i,list):
            new_list.append(remove_element(i,elem))
        else:
            if i!=elem:
                new_list.append(i)
    return new_list

nested_list = eval(input("Enter nested list: "))
print(nested_list)
element = int(input("Enter element you want to remove from the list: "))
print("removing element [",element,"] from the list")
result = remove_element(nested_list,element)
print(result)
```

```python
def find_max(lst):
    # max_num=lst[0] this gives error as list and int comparairiosn
    max_num = float("-inf")
    for i in lst:
        if isinstance(i,list):
            # max_num=find_max(i) this would override the last max number so add a sub max/temp max to check on previous max and new max and then give max to higher number
            sub_max = find_max(i)
            if sub_max>max_num:
                max_num = sub_max
        else:
            if i>max_num:
                max_num=i
    return max_num


nested_list = eval(input("Enter nested list: "))
print(nested_list)
result = find_max(nested_list)
print(result)
```

```python
def count_occurence(lst,elem):
    count=0
    for i in lst:
        if isinstance(i,list):
            count+=count_occurence(i,elem)
        else:
            if(i==elem):
                count+=1
    return count

nested_list = eval(input("Enter nested list: "))
print(nested_list)
element = int(input("Enter the element you want to count for: "))
result = count_occurence(nested_list,element)
print(result)
```

```python
def deep_flatten(lst):
    new_list = []
    for i in lst:
        if isinstance(i,list):
            # new_list.append(deep_flatten(i)) here append will add the entire list just as it is to the new list so it wont be a single list but still a nested list
            new_list.extend(deep_flatten(i)) #but extend will extend the new_list with the added list items so it will take the nested list and extend that into the existing list
        else:
            new_list.append(i)
    return new_list

nested_list = eval(input("Enter nested list: "))
print(nested_list)
result = deep_flatten(nested_list)
print(result)
```

```python
def sum_count(lst):
    sum=0
    count=0
    for i in lst:
        if isinstance(i,list):
            s,c = sum_count(i)
            sum+=s
            count+=c
        else:
            sum+=i
            count+=1
    return sum, count

def average_nested(lst):
    # sum = sum_count(lst)
    # count = sum_count(lst)
    sum,count = sum_count(lst)
    return sum/count

nested_list = eval(input("Enter nested list: "))
print(nested_list)
result = average_nested(nested_list)
print(result)
```

```python
# Creating two 3D arrays
arr1 = np.array([[[1], [2]], [[3], [4]]]) # Shape: (2, 2, 1)
arr2 = np.array([[[5], [6]], [[7], [8]]]) # Shape: (2, 2, 1)
# Concatenate along axis 0 (vertical stacking)
concat_axis0 = np.concatenate((arr1, arr2), axis=0)
# Shape of concat_axis0: (4, 2, 1), arr1 is stacked on top of arr2
# Concatenate along axis 1 (horizontal stacking)
concat_axis1 = np.concatenate((arr1, arr2), axis=1)
# Shape of concat_axis1: (2, 4, 1), arr1 and arr2 are placed side by side
# Concatenate along axis 2 (depth stacking)
concat_axis2 = np.concatenate((arr1, arr2), axis=2)
# Shape of concat_axis2: (2, 2, 2), arr1 and arr2 are stacked in depth (third
# dimension)
# Displaying the results
print("Concatenation along axis 0 (Vertical stacking):\n", concat_axis0)
print("Shape of concatenated array along axis 0:", concat_axis0.shape)
print("\nConcatenation along axis 1 (Horizontal stacking):\n", concat_axis1)
print("Shape of concatenated array along axis 1:", concat_axis1.shape)
print("\nConcatenation along axis 2 (Depth stacking):\n", concat_axis2)
print("Shape of concatenated array along axis 2:", concat_axis2.shape)
```

## Vectorization
converting an operation that uses loops into operation that uses Numpy array operations which optimizes performance. these are much faster than loop based equivalents.
### why vectorization
1. efficient = eliminates need of loop as numpy performs computation internally in compiled code
2. readability = code becomes more conside and easier to understand
3. memory efficiency = vectorized operations do not require creation of additional arrays as they work in place

this speeds up math operations by eliminating explicit loops and leverages optimized array computations.

```python
import time

size = 10**6 #** means power
a = np.random.rand(size)
b = np.random.rand(size)

start_time = time.time()
result_loop = np.zeros(size)
for i in range(size):
    result_loop[i] = a[i]+b[i]
end_time = time.time()
loop_time = end_time - start_time
print(f"loop execution time: {loop_time:.5f} seconds")


# using numpys vectorization
start_time = time.time()
result_vectorized = a+b
end_time = time.time()
vectorized_time = end_time - start_time
print(f"Vectorized time: {vectorized_time:.5f} seconds")
speedup = loop_time/vectorized_time
print(f"vectorization is approximately {speedup:.2f} times faster")
```

```python
#Element-wise Addition

size = 1000000

#Python lists
list1 = list(range(size))
list2 = list(range(size))
list_add = []

start = time.time()
for i in range(size):
    list_add.append(list1[i] + list2[i])
end = time.time()
print("Python list addition time:", end - start, "seconds")

#numpy arrays
arr1 = np.array(list1)
arr2 = np.array(list2)

start = time.time()
arr_add = arr1 + arr2
end = time.time()
print("NumPy array addition time:", end - start, "seconds")

#element wise multiplication

#python lists
list_mul = []
start = time.time()
for i in range(size):
    list_mul.append(list1[i] * list2[i])
end = time.time()
print("Python list multiplication time:", end - start, "seconds")

#numpy arrays
start = time.time()
arr_mul = arr1 * arr2
end = time.time()
print("NumPy array multiplication time:", end - start, "seconds")

# dot product
# Python lists
dot_list = 0
start = time.time()
for i in range(size):
    dot_list += list1[i] * list2[i]
end = time.time()
print("Python list dot product time:", end - start, "seconds")

# numpy arrays
start = time.time()
dot_arr = np.dot(arr1, arr2)
end = time.time()
print("NumPy array dot product time:", end - start, "seconds")

# matrix mult
matrix_size = 1000

# Python lists
py_mat1 = [[i for i in range(matrix_size)] for _ in range(matrix_size)]
py_mat2 = [[i for i in range(matrix_size)] for _ in range(matrix_size)]
py_result = [[0]*matrix_size for _ in range(matrix_size)]

start = time.time()
for i in range(matrix_size):
    for j in range(matrix_size):
        for k in range(matrix_size):
            py_result[i][j] += py_mat1[i][k] * py_mat2[k][j]
end = time.time()
print("Python lists matrix multiplication time:", end - start, "seconds")

# numpy arrays
np_mat1 = np.array(py_mat1)
np_mat2 = np.array(py_mat2)

start = time.time()
np_result = np_mat1 @ np_mat2
end = time.time()
print("NumPy arrays matrix multiplication time:", end - start, "seconds")
```
